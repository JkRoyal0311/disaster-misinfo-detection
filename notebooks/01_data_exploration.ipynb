```python
import pandas as pd
import sys
import os

# Allow the notebook to import from the src/ folder
sys.path.append(os.path.abspath('..'))
from src.preprocess import clean_text

# Load your provided dataset directly from the raw folder
df = pd.read_csv('../data/raw/fake_real_news_sample.csv')
print(f"Loaded {len(df)} samples from data/raw/fake_real_news_sample.csv")
df.head()
```

```python
# Apply our reusable cleaning function to the 'text' column
df['cleaned_text'] = df['text'].apply(clean_text)

# Encode labels for machine learning (real=0, fake=1)
df['label_encoded'] = df['label'].map({'real': 0, 'fake': 1})

# Save the processed data for Week 2
import os
os.makedirs('../data/processed', exist_ok=True)
df.to_csv('../data/processed/cleaned_disaster_news.csv', index=False)
print("Cleaned data saved to data/processed/cleaned_disaster_news.csv")

df[['cleaned_text', 'label_encoded']].head()
```
